# Competition: Credit Card Default Prediction

## The Task

Predict the probability that a credit card holder will **default on their next payment**.

You are given a dataset of credit card holders with their demographic information, credit limit, and 6 months of billing and payment history. Your job is to predict which customers will default.

## Data

| File | Description |
|---|---|
| `day2/generated/competition_train.csv` | ~21,000 rows with features + target (`default`) |
| `day2/generated/competition_test.csv` | ~9,000 rows — features only, no target |

## Column Descriptions

| Column | Description |
|---|---|
| `row_id` | Unique row identifier |
| `application_month` | Month of the credit application (YYYY-MM) |
| `employer_sector` | Applicant's employment sector |
| `limit_bal` | Credit limit (NT dollars) |
| `sex` | 1 = male, 2 = female |
| `education` | Education level |
| `marriage` | Marital status |
| `age` | Age in years |
| `pay_0` | Repayment status in the most recent month (-1 = on time, 1–9 = months of payment delay) |
| `pay_3` | Repayment status 3 months ago |
| `pay_6` | Repayment status 6 months ago |
| `bill_amt1` – `bill_amt6` | Monthly bill statement amount (1 = most recent, 6 = oldest) |
| `pay_amt1` – `pay_amt6` | Monthly payment amount (1 = most recent, 6 = oldest) |
| `months_since_first_credit` | Months since the customer's first credit account was opened |
| `num_credit_cards` | Number of active credit cards held by the customer |
| `etl_batch_id` | Internal ETL batch identifier |
| `schema_version` | Data schema version |
| **`default`** | **Target** — 1 = defaulted, 0 = did not default *(train only)* |

## Evaluation

**Metric: ROC-AUC** — we score your predicted probabilities against the true labels.

## Submission Format

A CSV with exactly two columns:

| row_id | predicted_probability |
|---:|---:|
| 0 | 0.073 |
| 1 | 0.891 |
| ... | ... |

- `row_id` must match the `row_id` in `competition_test.csv`
- `predicted_probability` is your model's P(default)
