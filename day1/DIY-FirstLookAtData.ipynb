### Give Me Some Credit

**Patterns commonly spotted**
- `RevolvingUtilizationOfUnsecuredLines` should be in [0, 1] but contains values > 5 000 (impossible / data error)
- `MonthlyIncome` has ~20% missing values; `NumberOfDependents` ~2.6% missing
- `DebtRatio` has extreme right-tail outliers
- Severe class imbalance: only ~6.7% positive (default) — ratio ~14:1
- Younger borrowers tend to have higher default rates
- Correlated delinquency features (30-59 / 60-89 / 90+ days past due)

### German Credit

**Kaggle Notebooks**
- [German Credit Analysis (EDA) — Plotly](https://www.kaggle.com/code/jiegzhan/german-credit-analysis-eda-plotly) — interactive visualizations, categorical breakdowns by risk class
- [German Credit Analysis — A Risk Perspective](https://www.kaggle.com/code/janiobachmann/german-credit-analysis-a-risk-perspective) — cost-sensitive framing (5:1 misclassification cost)
- [Predicting Credit Risk: EDA + Viz + Pipeline](https://www.kaggle.com/code/thiagopanini/predicting-credit-risk-eda-viz-pipeline) — full sklearn pipeline with preprocessing
- [German Credit Risk Analysis: Beginner's Guide](https://www.kaggle.com/code/pahulpreet/german-credit-risk-analysis-beginner-s-guide) — beginner-friendly walkthrough

**GitHub**
- [alexfierro9/German-Credit-EDA](https://alexfierro9.github.io/German-Credit-EDA/) — full pipeline from EDA to modeling

**Patterns commonly spotted**
- Checking account status is the strongest predictor of credit risk
- Longer loan durations and higher credit amounts correlate with higher default risk
- Certain loan purposes (education, vacation) are riskier than others (car purchases)
- Younger applicants tend to be classified as higher risk
- Low savings and unfavorable checking account status are strong indicators of bad credit
- Cost-sensitive evaluation matters: misclassifying a bad borrower as good costs 5x more than the reverse
