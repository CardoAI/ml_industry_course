Both are fetched directly from OpenML (no download or authentication required).

## Setup

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_openml

SEED = 42
np.random.seed(SEED)

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_colwidth", 200)

print("Environment ready.")
```

## Load Datasets

```python
# Give Me Some Credit (OpenML ID 45577)
gmsc_raw = fetch_openml(data_id=45577, as_frame=True, parser="auto")
gmsc = gmsc_raw.frame  # features + target in one DataFrame

print("Give Me Some Credit")
print(f"  Shape: {gmsc.shape}")
print(f"  Target: '{gmsc_raw.target.name}' — positive rate: {gmsc_raw.target.astype(int).mean():.4f}")
gmsc.head()
```

```python
# German Credit (OpenML ID 31)
german_raw = fetch_openml(data_id=31, as_frame=True, parser="auto")
german = german_raw.frame  # features + target in one DataFrame

print("German Credit")
print(f"  Shape: {german.shape}")
print(f"  Target: '{german_raw.target.name}' — class distribution:")
print(german_raw.target.value_counts())
german.head()
```
